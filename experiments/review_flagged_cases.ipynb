```python
import json
from pathlib import Path
from pprint import pprint

repo = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src/evaluation').is_dir())
folder = repo / 'output/judge_review_20261024'
notes_file = folder / 'draft_reviews.json'
notes = json.loads(notes_file.read_text(encoding='utf-8'))
records = json.loads((folder / 'inputs.json').read_text(encoding='utf-8'))

review_index = 0
item = notes['items'][review_index]
record = next(r for r in records if r['run_id'] == item['run_id'] and r['batch'] == item['batch'])
inputs = record['inputs']
print('Case', review_index + 1, 'of', len(notes['items']))
print(item['packet_id'], item['run_id'])
print('Fields checked:', ', '.join(item['scope']))
print('Targeted review:', notes.get('status', 'draft'))
print('Approved by:', notes['signoff'].get('approved_by', 'Not recorded'))
```

```python
for evidence in inputs['evidence']['evidence']:
    print('\n' + evidence['evidence_id'])
    print(' '.join(evidence['text'].split()))
print('\nFacts:')
pprint(inputs['evidence']['facts'])
print('\nInjected error:')
pprint(inputs['answer_key'])
```

```python
print('Initial report:')
pprint(inputs['initial_report'])
print('\nFinal report:')
pprint(inputs['final_report'])
print('\nReview:')
pprint(inputs['review'])
print('\nTrader:')
pprint(inputs['trader'])
```

```python
print('Accepted review note:')
print(item['note'])
for field, accepted in item['accepted_labels'].items():
    print(field, '| accepted review:', accepted, '| original judge:', item['judge_labels'][field])
print('\nAdditional personal check:')
pprint(item['human_check'])
```
